```python
import cv2
import matplotlib.pyplot as plt

figure = cv2.imread("Place the filename of an image here")

figure = cv2.cvtColor(figure, cv2.COLOR_BGR2RGB)

plt.imshow(figure)
plt.show()
```

```python
# Primeira modificação da primeira imagem
# First modification of the first image

# Fazendo as alterações nas camadas de cores das imagens. Na variável figure foi armazenada a imagem original.
# Making changes to the color layers of the images. The original image was stored in the figure variable.

figure_cp = figure.copy()
figure_cp[:,:,0] = figure[:,:,1]
figure_cp[:,:,1] = figure[:,:,2]
figure_cp[:,:,2] = figure[:,:,2]

# Exibindo o resultado das modificações feitas no passo anterior
# Displaying the result of the modifications made in the previous step

fig,axs = plt.subplots(1, 2, layout='constrained', figsize = [13,13])

axs[0].imshow(figure)

axs[0].set_title('Imagem original / Original Image')

axs[0].set_axis_off()

axs[1].imshow(figure_cp)                        ]

axs[1].set_title('Imagem modificada / Modified Image')

axs[1].set_axis_off()

plt.show()
```

```python
# Segunda modificação da primeira imagem
# Second modification of the first image

# Alterando as camadas de cores mais uma vez, de maneira análoga a que foi feita anteriormente. E o processo de exibição dos resultados
# é idêntico ao anterior.

# Changing the color layers once again, in a manner similar to what was done previously. And the process of displaying the results is
# identical to the previous one.

figure_cp = figure.copy()
figure_cp[:,:,0] = figure[:,:,0]
figure_cp[:,:,1] = figure[:,:,2]
figure_cp[:,:,2] = figure[:,:,1]

fig,axs = plt.subplots(1, 2, layout='constrained', figsize = [13,13])

axs[0].imshow(figure)

axs[0].set_title('Imagem original / Original Image')

axs[0].set_axis_off()

axs[1].imshow(figure_cp)

axs[1].set_title('Imagem modificada / Modified Image')

axs[1].set_axis_off()

plt.show()
```

```python
# Terceira modificação da primeira imagem
# Third modification of the first image

figure_cp = figure.copy()

figure_cp[:,:,0] = figure[:,:,1]
figure_cp[:,:,1] = figure[:,:,1]
figure_cp[:,:,2] = figure[:,:,2]

fig,axs = plt.subplots(1, 2, layout='constrained', figsize = [13,13])

axs[0].imshow(figure)

axs[0].set_title('Imagem original / Original Image')

axs[0].set_axis_off()

axs[1].imshow(figure_cp)

axs[1].set_title('Imagem modificada / Modified Image')

axs[1].set_axis_off()

plt.show()
```
